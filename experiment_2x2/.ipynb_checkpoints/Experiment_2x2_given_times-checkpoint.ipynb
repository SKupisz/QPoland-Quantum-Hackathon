```python
"""
Known x values substituted as constants: only t and b are variables.
f = min(x1,x3) + max(x1, x3, x2+x4)
"""
import numpy as np
from qiskit_optimization import QuadraticProgram
from qiskit_optimization.converters import QuadraticProgramToQubo

x1, x2, x3, x4 = 4, 2, 3, 3            # your numbers
M = 2                                    # penalty weight (> 1, see note below)

t_ub = max(x1, x3) + x2 + x4             # safe upper bound: max(x1,x3,x2+x4) <= max(x1,x3) + x2 + x4

qp = QuadraticProgram("minmax_fixed")
qp.integer_var(lowerbound=0, upperbound=t_ub, name="t")   # t ~ max(x1, x3, x2+x4)
qp.binary_var(name="b")                                   # b selects min(x1, x3)

# x3 + b*(x1 - x3) + t  with constants plugged in
qp.minimize(constant=x3, linear={"b": x1 - x3, "t": 1})

qp.linear_constraint({"t": 1}, ">=", x1,      "t_ge_x1")
qp.linear_constraint({"t": 1}, ">=", x3,      "t_ge_x3")
qp.linear_constraint({"t": 1}, ">=", x2 + x4, "t_ge_x2_plus_x4")

conv = QuadraticProgramToQubo(penalty=M)
qubo = conv.convert(qp)
print("qubits:", qubo.get_num_vars())

op, offset = qubo.to_ising()
print(op)
print("offset:", offset)

# exact check: Hamiltonian is diagonal, take its minimum
n = qubo.get_num_vars()
diag = np.real(op.to_matrix(sparse=True).diagonal()) + offset
idx = int(np.argmin(diag))
bits = np.array([(idx >> k) & 1 for k in range(n)])   # Qiskit is little-endian
sol = conv.interpret(bits)                            # [t, b]

f_true = min(max(x1 + x3, x1 + x2 + x4), max(x3 + x1, x3 + x2 + x4))
print("ground energy:", diag[idx], "| decoded (t, b):", sol, "| true f:", f_true)
```
